```python
import numpy as np
import torch 
import torch.nn as nn
from pooling import max_pooling

image = np.array([
        [1, 3, 2, 4],
        [5, 6, 1, 2],
        [7, 2, 8, 3],
        [4, 1, 9, 5]
    ])

result = max_pooling(image, pool_size=2, stride=2)

# expecting 
# [[6. 4.]
# [7. 9.]]
print(result)

# find y(2,1) , expecting 9.0
print(result[1,1])

# implement with torch
pool = nn.MaxPool2d(kernel_size=2, stride=2)
image = torch.tensor([
    [1, 3, 2, 4],
    [5, 6, 1, 2],
    [7, 2, 8, 3],
    [4, 1, 9, 5]
], dtype=torch.float32)

# unsqueeze() → adds a dimension
# squeeze() → removes a dimension 
# A CNN usually expects image input in this format:
#  (batch_size, channels, height, width)
# example (1, 1, 28, 28) is one image, one channel (grayscale), image of size 28x28
image = image.unsqueeze(0).unsqueeze(0)

output = pool(image)
print("Output:", output)

# output[batch, channel, i, j]
# get y(1, 0)
print(output[0, 0, 1, 0].item())
```
